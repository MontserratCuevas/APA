```python
class Clase:
    def __init__(self):
        self.valor = 0
    def incrementa(self):
        self.valor += 1
       # print(self)
    def __str__(self):
        return f'{self.valor}'
    def __iter__(self):
        return self
    def __next__(self):
        self.incrementa()
        if self.valor > 17:
            raise StopIteration
        return self.valor
```

```python
objeto = Clase()
```

```python
next(objeto)
```

```python
next(objeto)
```

```python
for o in objeto:
    print(o)
```

```python
def generator():
    valor = 0
    while True:
        valor += 8
        yield valor
        if valor > 50:
            return
```

```python
g = generator()
```

```python
g
```

```python
for o in g:
    print(o)
```
