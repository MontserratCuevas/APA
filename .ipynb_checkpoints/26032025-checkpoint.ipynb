Objeto.metodo(args)--> Clase metodo(objeto, args)

```python
1 .__add__(2)
```

```python
int.__add__(1,2)
```

```python
class Clase(int):
    def __add__(self, other):
        return self-other
#los parentesis en una clase sirve para definir el mecanismo de herencia
   
```

```python
numero = Clase(3)
```

```python
numero*2
```

```python
numero/2
```

```python
numero.__add__(1)
```

```python
numero + 4
```

```python
Clase.__add__ = int.__add__
```

```python
numero + 4
```

```python
Clase.__add__ = int.__pow__ #potencia
```

```python
numero + 4
```

```python
int.__add__ = int.__pow__ #no puedes modificar los tipos básicos
```

objeto._privado(...) es accesible sirve para avisar que no lo toquen
objeto.__muyprivado(...)

```python
class Clase():
    """
    Documentación de la clase que solo define dos atributos más o menos privados
    """
    _privado = 4
    __muy_privado = 24
```

```python
objeto = Clase
print(objeto._privado)
print(objeto._Clase__muy_privado)
```
