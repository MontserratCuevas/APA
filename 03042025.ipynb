````python
numeros = ["zero", "u", "dos", "tres"]
````

````python
NUMEROS = map(lambda s: s.upper(), numeros)
````

````python
#comprensión
NUMEROS2 = (s.upper() for s in numeros)
````

````python
for NUMERO in NUMEROS2: print(NUMERO)
````

````python
#comprensión
NUMEROS2 = (s.upper() for s in numeros if len(s)>3) #permite poner ifs
````

````python
for NUMERO in NUMEROS2: print(NUMERO)
````

filter(funcion, iterables...)
la funcion se va a tomar sus salidas como True o False
si es True se muestra por pantalla
si es False no se muestra en pantalla

````python
quintos = filter(lambda x: not x%5, range(40))
````

````python
next(quintos)
````

````python
for numero in quintos:
    print(numero)
````

````python
quintos = (x for x in range(40) if not x%5) # combinación de filter y map
````

````python
for numero in quintos:
    print(numero)
````

````python
from functools import reduce
````

````python
help(reduce)
````

````python
def factorial(n):
    return reduce(lambda x,y: x*y, range(2, n+1))
````

````python
factorial(10)
````

````python
for n in range(2,10):
    print(f"{n}! = {factorial(n)}")
````

````python
class Alumno:
    def __init__(self, nombre, apellido, nota):
        self.nombre, self.apellido, self.nota = nombre, apellido, nota
    def __str__(self):
        return f"{self.nombre} {self.apellido} {self.nota}"
    
    __repr__ = __str__

    def __call__(self, nota):
        self.nota = nota
````

````python
Alumnos = [Alumno("Tomàs", "Lloret",3), Alumno("Mark", "Bonete", 2), Alumno( "Montserrat", "Cuevas", 10)]
````

````python
sorted(Alumnos, key = lambda x: x.apellido)
````

````python
def ordenaAtributo(objetos, clave):
    return sorted(objetos, key = lambda objeto: objeto.__getattribute__(clave))
````

````python
ordenaAtributo(Alumnos,"nota")
````
