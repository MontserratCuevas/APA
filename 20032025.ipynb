```python
def funcion(uno, /, dos,*, tres ):
    print(uno, dos, tres)
funcion(1, dos = 2 , tres = 3)
#con * forzamos a que tres sea por clave, si no es por clave saldrá error
#con / forzamos que uno sea posicional
```

```python

def funcion2(*elementos, fin):
    print(elementos)
    #lista con todos los elementos posicionales no asignados
    for elemento in elementos:
        print(elemento)
    print(fin)
```

```python
funcion2(1, 2, 3, fin = 5) #te da una tupla
```

```python
def funcion(*lista, **diccionario):
    print(lista)
    print(diccionario)
```

```python
diccionario = {'u': 1, 'dos' : 2, 'tres':3, 'quatre' : 4}
funcion(**diccionario)
```
